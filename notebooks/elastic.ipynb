```python


# Chemins
s3_base_path = "s3://bkt-pud-uc/uc202-rex/FA3_useful_files_final"
csv_s3_path = "s3://bkt-pud-uc/uc202-rex/FA3_useful_files.csv"
local_base_dir = Path(r"\\atlas.edf.fr\CO\DIPNN-FLA3\Services.004")
upload_log_path = Path(r"C:\Users\e58883\uc202\uc202-ipn-rex\notebooks\uploaded_log.txt")  # local log file

# Lecture du fichier CSV
#df = PLE.s3.read_dataframe(csv_s3_path, file_format="csv")
df=pl.read_csv(r"C:\Users\e58883\Downloads\FA3_useful_files.csv")
print('read done')
# Vérifie la présence de la colonne attendue
if "Chemin d'accès" not in df.columns:
    raise ValueError("Colonne 'Chemin d'accès' manquante dans le fichier CSV.")

# Fichiers déjà uploadés (log)
if upload_log_path.exists():
    with open(upload_log_path, "r", encoding="MacRoman") as f:
        already_uploaded = set(line.strip() for line in f)
else:
    already_uploaded = set()


# Fonction d'upload avec log
def upload_file(file_path_str: str):
    file_path_str = file_path_str.replace("\\", "/").replace(":/", "://", 1)
    try:
        local_path_without_base = file_path_str.replace(
            "//atlas.edf.fr//CO/DIPNN-FLA3", ""
        )

# Construct the final S3 target path
        s3_target_path = f"{s3_base_path}{local_path_without_base}"
        PDE.s3.upload(local_file=file_path_str, path=s3_target_path)
        with open(upload_log_path, "a", encoding="utf-8") as log:
            log.write(Path(file_path_str).as_posix() + "\n")
        #return f"Uploaded: {file_path}"
    except Exception as e:
        return f"Error with {file_path_str}: {e}"


# Liste à uploader (exclure déjà uploadés)
all_paths = df["Chemin d'accès"].drop_nulls().unique().to_list()
files_to_upload = [p for p in all_paths if p not in already_uploaded]

del df

print(f"Total remaining to upload: {len(files_to_upload)}")

# Upload en parallèle
max_workers = 250
with ThreadPoolExecutor(max_workers=max_workers) as executor:
    futures = {
        executor.submit(upload_file, file_path): file_path
        for file_path in files_to_upload
    }

    for future in as_completed(futures):
        print(future.result())
```
