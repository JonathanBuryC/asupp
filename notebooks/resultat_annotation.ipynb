```python
import pandas as pd

# Lire le fichier Excel
file = pd.ExcelFile("annotation_autres.xlsx")

# Liste des colonnes communes
colonnes_communes = ["Question", "Doc ID", "RRF Score", "Content", "Annotation"]

# Initialiser une liste pour stocker les DataFrames de chaque feuille
dataframes = []

# Parcourir chaque feuille du fichier Excel
for sheet_name in file.sheet_names:
    # Lire la feuille actuelle
    df = pd.read_excel(file, sheet_name=sheet_name)

    # Sélectionner uniquement les colonnes communes
    df_commun = df[colonnes_communes]

    # Ajouter le DataFrame filtré à la liste
    dataframes.append(df_commun)

# Concaténer tous les DataFrames de la liste
resultat_autres = pd.concat(dataframes, ignore_index=True)

# Afficher le résultat
print(resultat_autres["Annotation"].notna().sum())
resultat_autres[resultat_autres["Annotation"].notna()].sort_values(by=["Question","Annotation"])
```

```python
resultat_autres.sort_values('Question')
```
