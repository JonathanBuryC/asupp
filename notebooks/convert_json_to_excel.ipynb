```python
import json
from openpyxl import Workbook # styles
# import pandas as pd
```

```python
file_path = "annotator_report_epria_boosted.json"
```

```python
data = None

with open(file_path) as file:
    data = json.load(file)
```

```python
# Create a sheet for the list of questions
wb = Workbook()

# Rename default sheet (usually named 'Sheet')
wb["Sheet"].title = "Liste des questions"

# Liste des questions
wb["Liste des questions"].append(["Numéro de question", "Question"])

for item in data:
    wb["Liste des questions"].append([item["question_id"], item["question"]])
wb["Liste des questions"].row_dimensions[1].height = None
```

```python
# Create a sheet for each question/answer pair

empty_line = ["", "", ""]
for item in data:

    sheet_id = item["question_id"]
    references = []
    # print(sheet_id)

    wb.create_sheet(sheet_id)

    wb[sheet_id].append(["Question", "", ""])
    wb[sheet_id].append(
        [
            item["question"],
            "",
            "",
        ]
    )
    wb[sheet_id].append(empty_line)
    wb[sheet_id].append(
        [
            "Réponse",
            "Annotation du paragraphe",
            "Commentaires",
        ],
    )

    if "llm_response_json" in item:
        try:
            if "error" in item["llm_response_json"]:
            
                print("error in llm response:")
        except:
            pass

        else:
            for i, theme in enumerate(item["llm_response_json"]):
                # print(theme)

                for j, subtheme in enumerate(theme["subthemes"]):

                    if j == 0:
                        wb[sheet_id].append([theme["theme"], "", "", ""])

                    wb[sheet_id].append(
                        [
                            f"{subtheme['subtheme']}\n\n{subtheme['content']}",
                            "",
                            "",
                        ]
                    )

                    references.extend(subtheme["references"])

    wb[sheet_id].append(empty_line)
    wb[sheet_id].append(
        [
            "Références dans la documentation",
            "Annotation de la référence",
            "Commentaires",
        ],
    )
    references = list(set(references))

    for chunk in item["chunks"]:
        if chunk["chunk_id"] in references:
            print(chunk["chunk_id"])
            wb[sheet_id].append(
                [
                    chunk["chunk_content"],
                    "",
                    "",
                ],
            )

    wb[sheet_id].column_dimensions["A"].width = 80
    #wb[sheet_id].column_dimensions["A"].alignment = styles.Alignment(       wrapText=True    )

#wb[sheet_id].row_dimensions[1].height = None
```

```python

# Save the workbook to a file
wb.save("annotator_report_epria_boosted.xlsx")
# Print a success message
print("Excel file created successfully!")
```
